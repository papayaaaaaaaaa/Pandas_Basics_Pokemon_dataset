**SERIES**

```python
%%capture
# gives us 2 columns -> INDEX and VALUES
data1 = ['A','B','C']
data2 = [100,102,104,106,108]
series1 = pd.Series(data1) # here S in Series is caital because this is a constructor and not a function
series2 = pd.Series(data2, index = ['a','b','c','d','e'])# If we want cuustom indexes, We can use List, Touple, Dictionary, numpy array, or another series also
# pd.series (list,index) OR pd.series(dictionary)
print (series1)
print()
print (series2)
```

```python
# int64 means that the datatype of the items of the elements is '64 bit integers'to help for internal type casting, no bit size for string because string its not necessary
```

```python
%%capture
print(series2.loc['a']) # loc['a'] means locate the value at label 'a'
print(series2.iloc[0]) # iloc[i] means locate the value at ith position (like array)
series1.loc[1] = 'N'   # to edit
```

```python
%%capture
# filtering by values
print(series2[(series2<105) & (series2>101)])
# && is a LOGICAL operator, & is a BITWISE operator (UNDERSTAND WHY LATER)
```

```python
%%capture
# USING A DICTIONARY
calories = {"Day 1": 1750, "Day 2": 2200, "Day 3": 1900}
CalorieSeries = pd.Series(calories) # there is no need for special indexing here because we are already using key-value pair of dict
print(CalorieSeries)
```

**DATAFRAME**

```python
%%capture
data = {'Name': ['Mayank','Arnab','Aaryan'], 'Age': [20,19,21]}
# keys give the key, Value gives the rest of the column
Employee = pd.DataFrame(data, index = ['Employee 1','Employee 2','Employee 3']) # if no index then we will have default indexing
# adding a new column
Employee['Job'] = ['Cook','Cashier','Doctor']
# datframe_name [key] = list of value for each column in a lsit
# adding a new row
new_rows = pd.DataFrame([{'Name': 'Reeyaan', 'Age': 22, 'Job': 'Pilot'} , {'Name': 'Ashwin', 'Age': 20, 'Job': 'Architect'}], index = ['Employee 4', 'Employee 5'])
# list of dictionaries for every row we want -> each dictionary -> key:value for all ab=vailable keys
Employee = pd.concat([Employee, new_rows]) # BEST WAY TO ADD A NEW ROW IS BY CONCATINATION -> takes in a list of dataframes to be added

print(Employee)
```

**FILTERING**

```python
%%capture
# we are creating a new dataframe of any pokemon have height more then 2
# we use Biitwise operators for comparisons (|,&) not logical operators (||, &&)
tall_pokemon = pokemon[pokemon['Height']>=2]
print(tall_pokemon)
print('\n-------------------------------------\n')
water_pokemon = pokemon[(pokemon['Type1']=='Water') | (pokemon['Type2']=='Water')]
print(water_pokemon)
print('\n-------------------------------------\n')
FireAndFlying_pokemon = pokemon[((pokemon['Type1']=='Fire') & (pokemon['Type2']=='Flying')) | ((pokemon['Type1']=='Flying') & (pokemon['Type2']=='Fire'))]
print(FireAndFlying_pokemon)
```

**AGGRAGATE FUNCTIONS**
- reduce the set of values to a single summary value (to summarise and analyze data)

```python
%%capture
print(pokemon.mean(numeric_only = True))  # 'nueric_only = True' means do not count the non numeric values 
print('-------------------------------------')
print(pokemon.sum(numeric_only = True))
print('-------------------------------------')
print(pokemon.min(numeric_only = True))
print('-------------------------------------')
print(pokemon.max(numeric_only = True))   
print('-------------------------------------')
print(pokemon.count(numeric_only = True)) # counts number of non null values 
```

```python
%%capture
# aggragate functions for particular column 
print(pokemon['Height'].mean())  # no need for the (Numeric_Value = True) because 'Height' coulmn has only numeric values
print('-------------------------------------')
print(pokemon[['Height','Weight']].mean())  # for multiple columns  
```

```python
%%capture
# creating GROUPS for the aggregate functions(if I want to look at the data of a particular group)
pokemon_Type1_Group = pokemon.groupby('Type1')
print(pokemon_Type1_Group[['Height','Weight']].mean())
```

**DATA CLEANING**

```python
%%capture
# (1) Dropping irrelevant columns [DROP THE HEIGHT AND WEIGHT COLUMNS] -> drop
pokemon_new = pokemon.drop(columns=['Height','Weight']) # name.drop(columns = list of columns to be dropped)
print(pokemon_new)
print('\n-------------------------------------\n')

# (2) To remove duplicate values
pokemon_new = pokemon.drop_duplicates()

# (3) Dropping any row that does not have a certain value in a given column [REMOVE ALL ROWS WHERE THE POKEMON HAS NO TYPE 2] -> dropna
pokemon_new = pokemon.dropna(subset = ['Type2']) # dropna removes any row where the data is missing
print(pokemon_new)
print('\n-------------------------------------\n') 

# (4) Replacing any missing values [REPLACE ANY MISSING VALUES IN TYPE 2 WITH 'NONE'] -> fillna
pokemon_new = pokemon.fillna({'Type2':'None'}) # fillna (dictionary where key is column and value is replacement of NaN term)
print(pokemon_new)
print('\n-------------------------------------\n')

# (5) Changing certain values in a column [MAKE ALL Water AND Fire IN TYPE 1 WATER AND FIRE
pokemon_new['Type1'] = pokemon['Type1'].replace({'Grass':'GRASS','Fire':'FIRE'}) 
# pokemon['Type1'] = pokemon['Type1'].replace({'Grass':'GRASS'})  ALSO WORKS IF I DO NOT WANT TO CREATE A NEW DATA FRAME
# new_name[column to be looked at] = name[column to be looked at].replace( dictionary where key is what we want to replace and value is what we want it to be replaced by )
print(pokemon_new)
print('\n-------------------------------------\n')

# (6) Changing all values to upper or to lower casae [CONVERT ALL THE TYPE 1 TO UPPER CASE]
pokemon_new['Type1'] = pokemon['Type1'].str.upper()
print(pokemon_new)
print('\n-------------------------------------\n')

# (7) Changing datatype of all data in given column [CHANGE THE LEGENDARY STATUS FROM 0,1 TO TRUE, FALSE]
pokemon_new['Legendary'] = pokemon['Legendary'].astype(bool)
print(pokemon_new)
```
